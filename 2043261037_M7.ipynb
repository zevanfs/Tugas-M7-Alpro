```python
#Nomor 1

bulan = input("Bulan apa Anda akan menginap? (Pilih salah satu: Mei, Juni, Juli, Agustus, September, Oktober)")
lama = int(input("Berapa lama Anda akan menginap? (Masukkan hanya dalam angka)"))

if (bulan == "Mei") or (bulan == "Oktober") :
    harga_studio = 500000
    harga_apartement = 650000
elif (bulan == "Juni") or (bulan == "September"):
    harga_studio = 752000
    harga_apartement = 687000
elif (bulan == "Juli") or (bulan == "Agustus"):
    harga_studio = 760000
    harga_apartement = 770000
else:
    print("Bulan yang anda masukkan tidak valid.")
    
harga_studio *= lama
harga_apartement *= lama

print("DATA MENGINAP DAN HARGA")
print(f"Bulan Menginap  : {bulan}")
print(f"Durasi Menginap : {lama} hari")

if lama > 3:
    diskon_studio = 0.15 * harga_studio
    diskon_apartement = 0.10 * harga_apartement
    harga_studio_diskon = harga_studio - diskon_studio
    harga_apartement_diskon = harga_apartement - diskon_apartement
    print(f"Studio       : Rp.{harga_studio_diskon:,}")
    print(f"Apartement  : Rp{harga_apartement_diskon:,}") 
else:
    print(f"Studio      : Rp.{harga_studio:,}")
    print(f"Apartement  : Rp.{harga_apartement:,}")
```

```python
#Nomor 5

nama_barang = input("Masukkan nama barang (Contoh: Pulpen):")
jumlah = float(input("Masukkan jumlah barang:"))
satuan_awal = input("Masukkan satuan awal (pcs/lusin/kodi):")
satuan_tujuan = input("Masukkan satuan tujuan (pcs/lusin/kodi):")

if satuan_awal == "pcs":
    total_pcs = jumlah
elif satuan_awal == "lusin":
    total_pcs = jumlah * 12
elif satuan_awal == "kodi":
    total_pcs = jumlah * 20
else:
    print("Satuan awal tidak dikenali.")
  
if satuan_tujuan == "pcs":
    hasil = total_pcs
elif satuan_tujuan == "lusin":
    hasil = total_pcs / 12
elif satuan_tujuan == "kodi":
    hasil = total_pcs / 20
else:
    print("Satuan tujuan tidak dikenali.")

print(f"Barang            : {nama_barang}")
print(f"Jumlah awal       : {jumlah} {satuan_awal}")
print(f"Total dalam pcs   : {total_pcs} pcs")
print(f"Konversi ke {satuan_tujuan} : {hasil} {satuan_tujuan}")
```
